Regla compuesta de Simpson.
Ejercicio 1 - Tarea 1.

 I= $\int_0^2 e^{(2x)}sen{(3x)} dx$

Para aproximar dentro $10^{(-4)}$

```python
import math
import matplotlib.pyplot as plt
```

```python
def f(x):
  """
  funcion del ej 1:
  f(x)=e^(2x)sen(3x)
  """
  return math.exp(2*x)*math.sin(3*x)
```

```python
a=0
b=2
n=66
```

Algoritmo 4.1

```python
h=(b-a)/n
```

```python
x0= f(a)+f(b)
```

$x_1=0$ suma de $f(x_i)$ para $i$ impar

$x_2=0$ suma de $f(x_i)$ para $i$ par

```python
x1=0
x2=0
```

```python
for i in range (1,n):
  X= a+i*h

  if i%2==0:
    x2=x2+f(X)
  else:
    x1=x1+f(X)

XI=h*(x0+2*x2+4*x1)/3

I_exacta=(math.exp(4)*(2*math.sin(6)-3*math.cos(6))+3)/13
```

Definimos la integral

```python
I=(math.exp(4)*(2*math.sin(6)-3*math.cos(6))+3)/13
```

 error absoluto

```python
error=abs(I-XI)
```

```python
print(f"n={n}")
print(f"h={h:.10f}")
print(f"aproximacion= {XI:.10f}")
print (f"error absoluto={error:.10e}")
```

Grafica de la funcion

```python
x_grafica=[
    a+i*(b-a)/1000
    for i in range (1001)
]
```

```python
y_grafica=[
    f(x)
    for x in x_grafica
]
```

```python
x_puntos=[
    a+i*h
    for i in range (n+1)
]

y_puntos=[
    f(x)
    for x in x_puntos
]
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_grafica,y_grafica,label=r"$f(x)e^{2x}\sin(3x)$", color="green")
plt.scatter(x_puntos,y_puntos,label="puntos de simpson", color="orange")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("regla compuesta de simpson")
plt.legend()
plt.grid(True)
plt.show()
```
